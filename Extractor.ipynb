```python
from lxml.html import fromstring
import os
import pandas as pd
```

```python
parser = fromstring(open("./pages/page_0.html").read())
```

```python
urls = parser.xpath('//*[@id="event-list-tab"]//lte-pfm-button/button/@data-url')
titles = parser.xpath('//*[@id="event-list-tab"]//lte-pfm-button/button/h3/text()')
dates = parser.xpath('//*[@id="event-list-tab"]//lte-pfm-button/button/div[1]/text()')
descriptions = parser.xpath('//*[@id="event-list-tab"]//lte-pfm-button/button/div[2]/text()')
```

```python
df = pd.DataFrame({
    "url": urls,
    "title": titles,
    "date": dates,
    "description": descriptions
})
```

```python
directory = 'pages'

df = pd.DataFrame(columns=["url", "title", "date", "description"])
for file in os.listdir(directory):
    with open(os.path.join(directory, file), 'r') as f:
        parser = fromstring(f.read())
        urls = parser.xpath('//*[@id="event-list-tab"]//lte-pfm-button/button/@data-url')
        titles = parser.xpath('//*[@id="event-list-tab"]//lte-pfm-button/button/h3/text()')
        dates = parser.xpath('//*[@id="event-list-tab"]//lte-pfm-button/button/div[1]/text()')
        descriptions = parser.xpath('//*[@id="event-list-tab"]//lte-pfm-button/button/div[2]/text()')

        df = pd.concat([df, pd.DataFrame({
            "url": urls,
            "title": titles,
            "date": dates,
            "description": descriptions
        })], ignore_index=True)
```

```python
df.to_csv('events_no_ics.csv', index=False, encoding='utf-8-sig')
```
