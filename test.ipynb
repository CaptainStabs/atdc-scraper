```python
from lxml.html import fromstring
import urllib
import re
```

```python
with open('overview.html', 'r') as f:
    html = f.read()
    parser = fromstring(html)


href = parser.xpath('//*[@id="event-navbar"]/div[1]/div[2]/div[1]/div[2]/div[2]/span/ul/li[4]/a/@href')[0]


parser.xpath('//*[@id="event-navbar"]/div[1]/div[2]/div[1]/div[2]/div[1]/div/span/text()')
```

```python
ics = href.split('/')[-1]

ics = urllib.parse.unquote(ics)
ics = ics.replace("&", '\n').replace("%20", " ")

start_match = re.search(r'startdt=(.+)', ics)
end_match = re.search(r'enddt=(.+)', ics)

startdt = start_match.group(1) if start_match else None
enddt = end_match.group(1) if end_match else None

print(startdt, enddt)

# print(ics.replace("&", '\n').replace("%20", " "))
```
